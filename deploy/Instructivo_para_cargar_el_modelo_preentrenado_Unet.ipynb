## 2. Cargar el modelo .h5

```python
from tensorflow.keras.models import load_model

# Cargar el modelo entrenado basado en ResNet50
modelo = load_model("modelo_16_Unet_PlantaEstado_Balanceado_Sin_Data_Aug_100perc.h5")

# Revisar la arquitectura (opcional)
modelo.summary()
```

## 3. Preparación de los datos de entrada
El modelo espera imágenes de tamaño 150x150 píxeles, con 3 canales RG

```python
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array

img_path = "ejemplo.jpg"
img = load_img(img_path, target_size=(128, 128))
x = img_to_array(img) / 255.0
x = np.expand_dims(x, axis=0)
```

## 4. Realizar una predicción

```python
# Obtener predicción
predicciones = modelo.predict(x)
y_pred = np.argmax(predicciones, axis=1)[0]
print(y_pred)
```

```python
# Cargar el archivo .json con la lista de indice a la clase de prediccion
import json

# Cargar el diccionario desde el JSON
with open("index_to_class_modelo_16_Unet_PlantaEstado_Balanceado_Sin_Data_Aug_100perc.json", "r") as f:
    index_to_class = json.load(f)

# Asegurarse de que las claves sean enteros
index_to_class = {int(k): v for k, v in index_to_class.items()}
```

```python
predicted_class_name = index_to_class[y_pred]
print(f"La clase predicha es: {predicted_class_name}")
```

### 4.1. Generación archivo JSON con predicción

```python
# Descomposición de clase
if "_" in predicted_class_name:
    planta, condicion = predicted_class_name.split("_", 1)
else:
    planta, condicion = predicted_class_name, None

# Crear resultado en formato JSON
result = {
    "archivo": img_path,
    "prediccion_clase": predicted_class_name,
    "Planta": planta,
    "Condición": condicion
}

# Guardar a JSON
with open("resultado_imagen.json", "w") as f:
    json.dump(result, f, indent=4, ensure_ascii=False)

# Imprimir por consola también si lo deseas
print(json.dumps(result, indent=4, ensure_ascii=False))
```
