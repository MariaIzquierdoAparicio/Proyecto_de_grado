Datos Balanceados por enfermedad

```python
import os
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Ruta de la carpeta
directory_root = Path("~/Documents/MFIA - PUJ - Maestría/Proyecto de Grado/0. Data/Datos por Estado Balanceados").expanduser()

# Obtener la lista de carpetas, ignorando archivos ocultos
carpetas = [f for f in directory_root.iterdir() if f.is_dir()]

# Lista para almacenar los datos
data = []

# Recorrer cada carpeta
for carpeta in carpetas:
    archivos = list(carpeta.glob("*"))  # Obtener todos los archivos dentro de la carpeta
    
    if archivos:  # Verificar si hay imágenes en la carpeta
        data.extend([[carpeta.name, archivo.name] for archivo in archivos])

# Crear el DataFrame
df = pd.DataFrame(data, columns=["Carpeta", "Nombre del archivo"])

# Separar la columna "Carpeta" en "Planta" y "Estado"
df[["Planta", "Estado"]] = df["Carpeta"].str.split("__", expand=True)

# Crear la columna "Saludable"
df["Saludable"] = np.where(df["Estado"] == "healthy", "Si", "No")

# Guardar en CSV
df.to_csv("archivos_datos_balanceados_estado1.csv", index=False)
```

```python
# Mostrar estadísticas generales
print("Resumen del DataFrame:")
print(df.describe(include="all"))

# Mostrar distribución de datos
print("\nFrecuencia de cada planta:")
print(df["Planta"].value_counts())

#print("\nFrecuencia de cada condición:")
#print(df["Condición"].value_counts())

print("\nConteo de imágenes saludables vs. no saludables:")
print(df["Saludable"].value_counts())

# Muestra aleatoria de datos
print("\nMuestra de 10 filas aleatorias:")
print(df.sample(10, random_state=42))
```
