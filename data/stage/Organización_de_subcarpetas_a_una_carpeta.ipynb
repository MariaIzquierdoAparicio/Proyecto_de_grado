```python
import os
import shutil
```

```python
# Carpeta de origen con subcarpetas
carpeta_origen = 'Datos por Estado Balanceados'

# Carpeta de destino donde se copiarán todas las imágenes
carpeta_destino = 'Datos Finales Estado Balanceados'

# Crear carpeta destino si no existe
os.makedirs(carpeta_destino, exist_ok=True)

# Recorrer recursivamente todos los archivos en la carpeta de origen
for root, dirs, files in os.walk(carpeta_origen):
    for archivo in files:
        # Filtrar por archivos de imagen (puedes añadir más extensiones si es necesario)
        if archivo.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
            ruta_origen = os.path.join(root, archivo)
            ruta_destino = os.path.join(carpeta_destino, archivo)

            # Renombrar si hay conflictos de nombre
            contador = 1
            nombre_base, extension = os.path.splitext(archivo)
            while os.path.exists(ruta_destino):
                nuevo_nombre = f"{nombre_base}_{contador}{extension}"
                ruta_destino = os.path.join(carpeta_destino, nuevo_nombre)
                contador += 1

            shutil.copy2(ruta_origen, ruta_destino)
```
